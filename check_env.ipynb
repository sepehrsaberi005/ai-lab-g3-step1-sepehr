## ۲. آزمون رسم نمودار با Matplotlib
برای اطمینان از صحت عملکرد موتور گرافیکی و نمایش صحیح نمودارها در محیط نوت‌بوک:

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 3.5))
plt.plot(x, y, label="sin(x)", color="#4f46e5", linewidth=2.5)
plt.title("Matplotlib Verification Test - Urban Route Finding", fontsize=12)
plt.xlabel("Time / Step")
plt.ylabel("Signal")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

## ۳. آزمون گراف شبکه معابر شهری با NetworkX
ساخت یک نمونه اولیه از تقاطع‌های شهری و معابر متصل‌کننده به همراه وزن مسافت (کیلومتر):

```python
G = nx.Graph()
G.add_edge("A (Meydan Azadi)", "B (Enghelab Sq)", weight=3.2)
G.add_edge("B (Enghelab Sq)", "C (Valiasr Sq)", weight=1.8)
G.add_edge("A (Meydan Azadi)", "D (Tohid Sq)", weight=2.4)
G.add_edge("D (Tohid Sq)", "C (Valiasr Sq)", weight=2.9)
G.add_edge("C (Valiasr Sq)", "E (Tajrish Sq)", weight=6.5)

pos = nx.spring_layout(G, seed=42)
plt.figure(figsize=(8, 5))
nx.draw(G, pos, with_labels=True, node_color="#818cf8", node_size=2400, 
        font_size=9, font_weight="bold", edge_color="#94a3b8", width=2.5)
edge_labels = nx.get_edge_attributes(G, "weight")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=10)
plt.title("Sample Urban Intersection Network (NetworkX Initial Test)", fontsize=12)
plt.axis("off")
plt.show()
```
