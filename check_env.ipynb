### ۲. بررسی نسخه کتابخانه‌های پایه علمی و محاسباتی

### ۳. تست عملیاتی رسم نمودار با Matplotlib و ساخت گراف با NetworkX

```python
import matplotlib.pyplot as plt

# ساخت یک گراف ساده شهری برای تست صحت عملکرد NetworkX و Matplotlib
G = nx.grid_2d_graph(3, 3)
pos = {(x, y): (x, y) for x, y in G.nodes()}

plt.figure(figsize=(5, 5))
nx.draw(G, pos=pos, with_labels=True, node_color='#38bdf8', node_size=600, edge_color='#64748b')
plt.title('Sanity Check: 3x3 Street Grid Test')
plt.show()

print('محیط کاری پایتون و تمام کتابخانه‌ها بدون مشکل آماده اجرا هستند.')
```
