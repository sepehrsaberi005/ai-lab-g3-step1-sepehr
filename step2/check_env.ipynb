```python
# بررسی نسخه پایتون و ماژول‌های اصلی
import sys
import os
import json

print("Python version:", sys.version)
assert sys.version_info >= (3, 9), "نیازمند پایتون نسخه 3.9 یا بالاتر هستید."

import numpy as np
print("NumPy version:", np.__version__)

import pandas as pd
print("Pandas version:", pd.__version__)

import matplotlib
print("Matplotlib version:", matplotlib.__version__)

import heapq
from collections import deque
print("Heapq & Deque: در دسترس هستند.")
```

```python
# بررسی بارگذاری داده‌های نقشه از پوشه maps
map_path = os.path.join("maps", "sample_city_map.json")
assert os.path.exists(map_path), f"فایل نقشه {map_path} یافت نشد!"

with open(map_path, "r", encoding="utf-8") as f:
    data = json.load(f)

grid = np.array(data.get("grid", data.get("grid_obstacles")))
traffic = np.array(data.get("traffic", data.get("traffic_costs")))

print(f"ابعاد نقشه بارگذاری‌شده: {grid.shape[0]} سطر در {grid.shape[1]} ستون")
print(f"محدوده وزن‌های ترافیکی معابر: حداقل {traffic[grid==0].min()} و حداکثر {traffic[grid==0].max()}")
print("✅ تمام نیازمندی‌های محیطی و داده‌های گام دوم با موفقیت تأیید شدند.")
```
